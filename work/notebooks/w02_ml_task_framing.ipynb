## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule would be too simple because content performance depends on several signals at the same time. A page can have low CTR but still receive strong impressions, or have an older update date but show improving trends. Different content types and intents can also behave differently. ML can learn combinations of these signals and produce a consistent score that helps prioritize pages instead of relying on one fixed threshold.
